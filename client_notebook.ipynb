# VMware VM Socket Client (Jupyter Notebook Version)

此 Notebook 用於在**本機電腦**上運行，連線至 VMware 虛擬機中的 Socket 伺服器以進行測試。

### 1. 連線至虛擬機
請在下方修改 `VM_IP` 為您虛擬機的實際 IP，然後執行 Cell 發起連線。

```python
import socket

# 1. 請在此修改為您虛擬機的實際 IP
VM_IP = "192.168.123.4" 
PORT = 7777

# 2. 建立 TCP Socket
client_socket = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
client_socket.settimeout(5.0)

try:
    print(f"正在連線至虛擬機 {VM_IP}:{PORT}...")
    client_socket.connect((VM_IP, PORT))
    print("[+] 連線成功！")
    
    # 3. 發送測試訊息
    msg = "Hello from Jupyter Notebook on Local Host!"
    print(f"[->] 發送訊息: '{msg}'")
    client_socket.sendall(msg.encode('utf-8'))
    
    # 4. 接收伺服器回覆
    data = client_socket.recv(1024)
    print(f"[<-] 收到伺服器回覆:\n{data.decode('utf-8')}")
    
except socket.timeout:
    print("[!] 連線逾時，請確認虛擬機 IP、背景 Server 是否開啟，以及網路/防火牆設定。")
except Exception as e:
    print(f"[!] 發生錯誤: {e}")
finally:
    # 5. 關閉連線
    client_socket.close()
    print("[*] 連線已關閉。")
```
